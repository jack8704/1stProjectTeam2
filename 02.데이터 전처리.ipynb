```python
# 식별자를 제외하도록 함수 수정
def check_outlier(df):
    # 수치형 컬럼 선택
    numeric_cols = df.select_dtypes(include='number').columns # df에서 숫자형(int, float 등) 데이터 타입을 가진 컬럼만 선택

    # 이상치 분석에서 제외할 식별자 컬럼 지정 -> exclude_cols =[제외할 컬럼 지정]
    # 연도/분기, 상권, 지역 등을 구분하기 위한 "코드(식별번호)"이므로 제외
    exclude_cols = [
        '기준년분기코드',
        '상권코드',
        '상권배후지코드',
        '서울시코드',
        '자치구코드',
        '행정동코드'
    ]

    # 실제로 분석할 숫자형 컬럼만 남기기 = 숫자형 컬럼 중에서 식별자 컬럼 제외 작업
    numeric_cols = [
        col for col in numeric_cols # numeric_cols에 있는 컬럼을 하나씩 col에 넣어서 확인.
        if col not in exclude_cols # 현재 컬럼이 exclude_cols(제외할 식별자 컬럼 명단) 안에 없다면 남김.
    ]

    # 결과를 저장할 빈 리스트 생성 -> result
    result = []

    for col in numeric_cols: # 컬럼을 하나씩 검사
        Q1 = df[col].quantile(0.25) # 현재 검사중인 컬럼을 가져옴 -> 해당 데이터의 25% 지점 값을 구함.   # 1사분위수(Q1)
        Q3 = df[col].quantile(0.75) # 현재 검사중인 컬럼을 가져옴 -> 해당 데이터의 75% 지점 값을 구함.   # 3사분위수(Q3)
        IQR = Q3 - Q1 # IQR = 3사분위수 - 1사분위수  # IQR = 통계 용어 "InterQuartile Range(사분위 범위)"의 약자.

        lower = Q1 - 1.5 * IQR # 이상치 하한값 계산
        upper = Q3 + 1.5 * IQR # 이상치 상한값 계산
        # 여기서 1.5는 데이터에서 일반적인 변동 범위를 적절하게 잡기 위한 실용적인 기준.
        
        
        outlier_count = (
            (df[col] < lower) |  # 하한값을 실제 데이터와 비교하여 너무 작은 값 찾기.
            (df[col] > upper)   #  상한값을 실제 데이터와 비교하여 너무 큰 값 찾기.
        ).sum() # 이상치 후보는 총 몇개인가?

        result.append([
            col, # 현재 검사 중인 컬럼(col)을 append하여 리스트 맨 뒤에 검사해서 또 추가함.
            outlier_count,
            lower,
            upper
        ])

    return pd.DataFrame(
        result,
        columns=['컬럼명', '이상치후보수', '하한값', '상한값']
    )
```

```python
print("=== 점포-상권 ===")
display(check_outlier(store_clean))
print("\n=== 상권변화지표-상권 ===")
display(check_outlier(change_clean))
print("\n=== 2023년-추정매출-상권 ===")
display(check_outlier(sales_2023))
print("\n=== 2024년-추정매출-상권 ===")
display(check_outlier(sales_2024))
print("\n=== 2025년-추정매출-상권 ===")
display(check_outlier(sales_2025))
print("\n=== 추정매출-상권배후지 ===")
display(check_outlier(sales_bg_clean))
print("\n=== 길단위인구-상권 ===")
display(check_outlier(floating_clean))
print("\n=== 상주인구-상권 ===")
display(check_outlier(resident_trade_clean))
print("\n=== 상주인구-상권배후지 ===")
display(check_outlier(resident_bg_clean))
print("\n=== 상주인구-서울시 ===")
display(check_outlier(resident_seoul_clean))
print("\n=== 상주인구-자치구 ===")
display(check_outlier(resident_gu_clean))
print("\n=== 상주인구-행정동 ===")
display(check_outlier(resident_dong_clean))
print("\n=== 영역-상권 ===")
display(check_outlier(area_clean))
```
