# 결측치 확인 및 처리

```python
# 결측치 확인 및 처리
def missing_check(df):
    result = pd.DataFrame({
        "결측치": df.isnull().sum(),
        "결측률(%)": (df.isnull().mean() * 100).round(2)
    })

    return result[result["결측치"] > 0]

display(missing_check(store_clean))
display(missing_check(change_clean))
display(missing_check(sales_clean))
display(missing_check(sales_bg_clean))
```

```python
# 식별자를 제외하도록 함수 수정
def check_outlier(df):
    # 수치형 컬럼 선택
    numeric_cols = df.select_dtypes(include='number').columns

    # 이상치 분석에서 제외할 식별자
    exclude_cols = [
        '기준년분기코드',
        '상권코드'
    ]

    numeric_cols = [
        col for col in numeric_cols
        if col not in exclude_cols
    ]

    result = []

    for col in numeric_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outlier_count = (
            (df[col] < lower) |
            (df[col] > upper)
        ).sum()

        result.append([
            col,
            outlier_count,
            lower,
            upper
        ])

    return pd.DataFrame(
        result,
        columns=['컬럼명', '이상치후보수', '하한값', '상한값']
    )
```

```python
print("=== 점포-상권 ===")
display(check_outlier(store_clean))
print("\n=== 상권변화지표-상권 ===")
display(check_outlier(change_clean))
print("\n=== 추정매출-상권 ===")
display(check_outlier(sales_clean))
print("\n=== 추정매출-상권배후지 ===")
display(check_outlier(sales_bg_clean))
```
